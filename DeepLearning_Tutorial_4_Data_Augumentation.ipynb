**Part 1:**

**Understanding of Part 1**

Tutorial 4 teaches data augmentation with Keras. You take one image and use ImageDataGenerator to produce many randomly altered versions of it (rotated, sheared, zoomed, flipped, brightness-changed). This increases the effective size of a dataset and adds variability, which is useful when collecting thousands of images isn't practical.


**Step 1: Importing Necessary Libraries**

```python
# Import necessary libraries
import matplotlib.pyplot as plt  # For displaying images
from tensorflow.keras.preprocessing.image import ImageDataGenerator  # For image augmentation
from tensorflow.keras.utils import img_to_array, load_img  # For loading and converting images to arrays
import os  # For creating directories and handling file system operations
```

**Step 2: Define the Folder Where Augmented Images Will Be Saved**

```python
# Define the folder where augmented images will be saved
save_folder = r"D:\Deep Learning Course\Codes\Augmented Images"

# Create the folder if it doesn't exist
os.makedirs(save_folder, exist_ok=True)
```

**Step 3: Initialize the ImageDataGenerator**

```python
# Initialize the ImageDataGenerator class with augmentation parameters
datagen = ImageDataGenerator(
    rotation_range=40,  # Randomly rotate images by up to 40 degrees
    shear_range=0.2,  # Apply random shearing (distorting the shape)
    zoom_range=0.2,  # Randomly zoom in or out on the image by 20%
    horizontal_flip=True,  # Randomly flip images horizontally
    brightness_range=(0.5, 1.5))  # Randomly adjust brightness between 50% and 150%
```

**Step 4: Load the Image and Convert It to a NumPy Array**

```python
# Load the sample image from the given path (replace with your own image path)
img = load_img(r"/content/flower image.avif")

# Convert the loaded image into a NumPy array
x = img_to_array(img)
```

**Step 5: Reshape the Image for Batch Processing**

```python
# Reshape the image to add an extra dimension for batch processing
x = x.reshape((1,) + x.shape)
```

**Step 6: Generate and Save Augmented Images**

```python
# Initialize a counter to limit the number of augmented images generated
i = 0

# Loop to generate and save augmented images
for batch in datagen.flow(x, batch_size=1,
                          save_to_dir=save_folder,  # Directory to save augmented images
                          save_prefix='image',  # Prefix for saved image filenames
                          save_format='jpeg'):  # Format to save the images in
    # Convert the batch back to an image (for visualization)
    augmented_image = batch[0].astype('uint8')  # Convert to an unsigned 8-bit integer format

    # Display the augmented image using matplotlib (optional)
    plt.figure()  # Create a new figure for each image
    plt.imshow(augmented_image)  # Show the augmented image
    plt.axis('off')  # Hide axes for better visualization
    plt.show()  # Display the image

    # Increment the counter
    i += 1

    # Stop after generating and saving 20 images
    if i > 20:
        break
```

**Step 7: Confirm the Saved Location**

```python
# Print the location of saved augmented images
print(f"Augmented images are saved in the folder: {os.path.abspath(save_folder)}")
```
