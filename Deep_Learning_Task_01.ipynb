#Tutorial 1-Task 1

The perceptronf.ipynb solves for the following task. It is related to the understanding of a basic perceptron. We are working with the IRIS dataset: https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data

TASK 1: Take manual input for the four features (sepal length, sepal width, petal length, and petal width).

Use the model to predict whether it's Iris-setosa or not based on the manually provided values.
Display the prediction result.

```python
import pandas as pd
#pandas is a python library for working with tables of data.
#In this case, it is a table of inputs (features) of the flowers and a label that indicates the species of the flower. Our dataset
#pandas as pd means, form now on in the rest of the script, the pandas library will be referred to as pd

from sklearn.utils import shuffle
#sklearn is for scikit-learn, a python machine learning library. We are importing a function wihtin this library, called "shuffle"
#We are randomly rearranging rows of the dataset. We do this because we later divide the data into training and testing sets. So randomizing the order

import numpy as np
#numpy is another python library. used for numerical calculations and arrays

from sklearn.linear_model import Perceptron
#here we are importing a pre-built perceptron model from scikit-learn
```

```python
df = pd.read_csv(
    'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data',
    header=None
)
#calling pandas (pd) to read the iris dataset and storing the resulting table/data in a variable called "df"
#df is just a varibale name, normally stands for dataframe. we can use any other name as well.
#this dataset doesn't have column names (headings), hence we use header=None,
#so that column headings are numerically set as 0, 1, 2, 3, 4 (0,1,2, &3) are the features/inuts and 4 will be the lables of the flowers
#this dataset has 3 classes or 3 species of flowers,which is what the perceptron will learn to predict

df = shuffle(df)
#Take df, shuffle its rows, and put the shuffled version back into df. we are introducing randomness
```

```python
df.head()
#visualize/show me first few rows of the data
```

```python
X = df.iloc[:, 0:4].values #x= inputs/features
#X takes four columns; 0 through 3. This makes a matric of the inputs/features. the dataset we have.

Y = df.iloc[:, 4].values #y = labels/outputs/answers
#Y takes column 4, i.e., the labels or the actual answer. since it only one coumn, so we write [:,4].
#the model/perceptron will get X and will try to learn how to arrive at the answers in Y (prediction)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
train_data, test_data, train_labels, test_labels = train_test_split (X, Y, test_size=0.25)
#we are now splittng out dataset into train and test groups
#25% = testig, rest of the 75% = training
```

```python
perceptron = Perceptron(eta0=0.1, max_iter=50)
#now we are creating the perceptron model, by using the already built-in model from scikit-learn, but by setting some parameters ourselves
#this is the learning rate: eta0=0.1 (η (eta) is controlling how quickly the model learns)
#max_iter=50 means the model can go through the training process up to 50 times
```

```python
perceptron.fit(train_data, train_labels)
#fit() means train the model. Take my data from 'train_data' and the corresponding labels from 'train_labels' and learn from them.
#during this process, the perceptron adusts its internal weights too
#since its a built-n model from scikit-learn, we dont need to worry about writing weight adjustment equations ourselves
```

```python
test_preds = perceptron.predict(test_data)
#now that the model has learnt by going through the encessary learning iterations too (as we defined above),
#it is time to test its prediction accuracy on the 'test_data'
```

```python
y_preds = perceptron.predict(test_data)
accuracy = accuracy_score(y_preds, test_labels)
#then we use the accuracy function to calculate accuracy of predictions
#this compares model precitions with actual answers

print('Accuracy:', round(accuracy, 2) * 100, "%")
print (test_preds)
#displaying the accuracy. We may see different accuracies if we repeat the run code mulitple times since we are shuffling and introducing randomness
```

```python
#for task 1 in tutorial 1 (it doesnt ask us to predict all 3 classes of the flower species. Rather if it is Iris setosa or not
#hence, now we solve for that
#we manually ask user to provide inputs, i.e., features of the flower, sepal length and width and petal length and width

sepal_length = float(input("Enter sepal length: "))
sepal_width = float(input("Enter sepal width: "))
petal_length = float(input("Enter petal length: "))
petal_width = float(input("Enter petal width: "))

# Now we will put the four inputs into one row
manual_input = [[
    sepal_length,
    sepal_width,
    petal_length,
    petal_width
]]

#asking perceptron to make prediction
prediction = perceptron.predict(manual_input)[0]

# printing/visualizing result
if prediction == 1:
    print("Prediction: Iris-setosa")
else:
    print("Prediction: Not Iris-setosa")
```
